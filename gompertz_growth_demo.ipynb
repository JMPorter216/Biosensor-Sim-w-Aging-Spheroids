```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
class GompertzGrowth:
    """
    Implements the Gompertz growth model.
    """

    def __init__(self, a: float, b: float, c: float):
        """
        Initializes the Gompertz growth model with parameters.

        Parameters:
        a (float): The upper asymptote.
        b (float): The displacement along the x-axis.
        c (float): The growth rate.
        """
        self.a = a
        self.b = b
        self.c = c

    def evaluate(self, t: np.ndarray) -> np.ndarray:
        """
        Evaluates the Gompertz function at given time points.

        Parameters:
        t (np.ndarray): Array of time points.

        Returns:
        np.ndarray: Evaluated values of the Gompertz function.
        """
        return self.a * np.exp(-self.b * np.exp(-self.c * t))
    
    def plot(self, t: np.ndarray) -> None:
        values = self.evaluate(t)
        plt.plot(t, values)
        plt.title("Spheroid Growth vs. Time (Gompertz Model)")
        plt.xlabel("Time")
        plt.ylabel("Growth")
        plt.show()
```

```python
model = GompertzGrowth(a=1.0, b=2.0, c=0.5)
time_points = np.linspace(0, 10, 100)

# Evaluate and print growth values
growth_values = model.evaluate(time_points)
print(growth_values)

# Plot the growth curve
model.plot(time_points)
```

# Gompertz Growth Model

This notebook demonstrates the Gompertz growth model for spheroids.

- `GompertzGrowth` class implements the growth function.
- `evaluate(t)` computes growth values.
- `plot(t)` visualizes the growth curve.

You can adjust parameters `a`, `b`, and `c` to explore different growth behaviors.


## Example Usage

We create a `GompertzGrowth` instance with parameters:
- `a = 1.0` (upper asymptote)
- `b = 2.0` (displacement)
- `c = 0.5` (growth rate)

We then plot the growth curve over time.


```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider

def gompertz(a, b, c):
    t = np.linspace(0, 10, 200)
    y = a * np.exp(-b * np.exp(-c * t))

    plt.figure(figsize=(6,4))
    plt.plot(t, y)
    plt.title("Gompertz Growth Model — Interactive")
    plt.xlabel("Time")
    plt.ylabel("Growth")
    plt.grid(True)
    plt.show()

interact(
    gompertz,
    a=FloatSlider(min=0.1, max=5.0, step=0.1, value=1.0, description="a"),
    b=FloatSlider(min=0.1, max=5.0, step=0.1, value=2.0, description="b"),
    c=FloatSlider(min=0.05, max=2.0, step=0.05, value=0.5, description="c"),
)
```
